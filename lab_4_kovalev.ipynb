## Описание датасета
Для анализа выбрана таблица "Common player info" из "NBA Database" - набор данных о игроках NBA, содержащий информацию о параметрах, карьере и истории игроков.

**Источник:** Kaggle, "NBA Database" : https://www.kaggle.com/datasets/wyattowalsh/basketball

**Состав набора данных:**

- **person_id** — уникальный идентификатор игрока в базе данных NBA.  
- **first_name** — имя игрока.  
- **last_name** — фамилия игрока.  
- **display_first_last** — отображаемое имя игрока в формате «Имя Фамилия».  
- **display_last_comma_first** — отображаемое имя игрока в формате «Фамилия, Имя».  
- **display_fi_last** — сокращённое отображаемое имя игрока (первая буква имени и фамилия).  
- **player_slug** — текстовый идентификатор игрока (slug), используемый в URL и внутренних системах.  

- **birthdate** — дата рождения игрока.  
- **school** — учебное заведение (школа или университет), которое посещал игрок.  
- **country** — страна происхождения игрока.  
- **last_affiliation** — последнее место спортивной или учебной принадлежности игрока (школа, колледж, клуб).  

- **height** — рост игрока (как правило, в строковом формате, например, «6-8»).  
- **weight** — вес игрока (в фунтах).  
- **season_exp** — количество сезонов профессионального опыта в NBA.  

- **jersey** — номер майки игрока.  
- **position** — игровая позиция игрока (например, Guard, Forward, Center).  
- **rosterstatus** — текущий статус игрока в составе команды (активен, неактивен и т.д.).  
- **games_played_current_season_flag** — флаг, указывающий, принимал ли игрок участие в текущем сезоне.  

- **team_id** — уникальный идентификатор команды, за которую выступает игрок.  
- **team_name** — название команды NBA.  
- **team_abbreviation** — аббревиатура названия команды.  
- **team_code** — код команды, используемый в информационных системах NBA.  
- **team_city** — город, который представляет команда.  

- **playercode** — внутренний код игрока, используемый в системе NBA.  

- **from_year** — год начала профессиональной карьеры игрока в NBA.  
- **to_year** — год окончания профессиональной карьеры игрока или последний активный сезон.  

- **dleague_flag** — флаг, указывающий, выступал ли игрок в G-League (Development League).  
- **nba_flag** — флаг, указывающий, связан ли игрок с NBA.  
- **games_played_flag** — флаг, указывающий, принимал ли игрок участие хотя бы в одной игре.  

- **draft_year** — год драфта NBA, в котором был выбран игрок.  
- **draft_round** — раунд драфта NBA, в котором был выбран игрок.  
- **draft_number** — номер пика на драфте NBA.  

- **greatest_75_flag** — флаг, указывающий, включён ли игрок в список 75 величайших игроков NBA.

## 2. Анализ пропусков

```python
missing_na = data.isna().sum()

missing_na = missing_na[missing_na > 0].sort_values(ascending=False)

print("Явные пропуски (NaN) по столбцам:")
print(missing_na)
```

```python
obj_cols = data.select_dtypes(include=["object"]).columns

data[obj_cols] = data[obj_cols].apply(lambda s: s.astype(str).str.strip())

missing_markers = [
    "", " ", "  ", "\t", "\n",
    "?", "??", "-", "--", "—",
    "NA", "N/A", "n/a", "NaN", "nan",
    "NULL", "null", "None", "none",
    "<NA>", "(null)", "missing", "Missing"
]

marker_counts = {}
for col in obj_cols:
    marker_counts[col] = data[col].isin(missing_markers).sum()

marker_counts = pd.Series(marker_counts).sort_values(ascending=False)
print("Скрытые пропуски (по маркерам) в object-колонках:")
print(marker_counts)

data[obj_cols] = data[obj_cols].replace(missing_markers, np.nan)

na_counts2 = data.isna().sum().sort_values(ascending=False)
na_pct2 = (na_counts2 / len(data) * 100).round(2)

summary_na2 = pd.DataFrame({"na_count": na_counts2, "na_%": na_pct2})
print("NaN после замены скрытых маркеров:")
print(summary_na2)
```

```python
num_cols = data.select_dtypes(include=[np.number]).columns

id_candidates = ["person_id", "team_id"]
name_candidates = ["display_first_last", "display_fi_last", "first_name", "last_name", "player_slug"]

id_col = next((c for c in id_candidates if c in data.columns), None)
name_col = next((c for c in name_candidates if c in data.columns), None)
base_cols = [c for c in [id_col, name_col] if c is not None]

nonpos_counts = (data[num_cols] <= 0).sum().sort_values(ascending=False)
nonpos_pct = (nonpos_counts / len(data) * 100).round(2)

summary = pd.DataFrame({
    "count_le_0": nonpos_counts,
    "pct_le_0": nonpos_pct
})

print("Сводка значений <= 0 по числовым столбцам:")
print(summary.loc[summary["count_le_0"] > 0])

suspects = {}
for col in num_cols:
    cond = data[col].isna() | (data[col] <= 0)
    if cond.sum() > 0:
        suspects[col] = data.loc[cond, base_cols + [col]].head(30)

for col, df in suspects.items():
    print(f"\n--- Примеры (NaN или <=0) для {col} ---")
    print(df)
```

```python
import matplotlib.pyplot as plt
miss_matrix = data.isna()

plt.figure(figsize=(12, 6))
plt.imshow(miss_matrix, aspect="auto", interpolation="nearest")
plt.title("Карта пропусков (NaN=True)")
plt.xlabel("Колонки")
plt.ylabel("Строки")
plt.xticks(range(len(data.columns)), data.columns, rotation=90)
plt.colorbar()
plt.tight_layout()
plt.show()
```

## 3. Статистические оценки

```python
numeric_cols = data.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = data.select_dtypes(include=['object']).columns.tolist()

print("Числовые признаки:", numeric_cols)
print("Категориальные признаки:", categorical_cols)

print("\n" + "="*50)
print("СТАТИСТИКИ ДЛЯ ЧИСЛОВЫХ ПРИЗНАКОВ")
print("="*50)

for col in numeric_cols:
    if col not in ['Championships']:
        print(f"\n{col}:")
        print(f"  Среднее: {data[col].mean():.2f}")
        print(f"  Медиана: {data[col].median():.2f}")
        print(f"  Стандартное отклонение: {data[col].std():.2f}")
        print(f"  Минимум: {data[col].min():.2f}")
        print(f"  Максимум: {data[col].max():.2f}")

print("\n" + "="*50)
print("СТАТИСТИКИ ДЛЯ КАТЕГОРИАЛЬНЫХ ПРИЗНАКОВ")
print("="*50)

for col in categorical_cols:
    if col not in ['Driver', 'Seasons', 'Championship Years']:
        print(f"\n{col}:")
        value_counts = data[col].value_counts()
        print(f"  Уникальных значений: {data[col].nunique()}")
        print(f"  Мода: {data[col].mode().iloc[0] if not data[col].mode().empty else 'Нет данных'}")
        print(f"  Распределение (топ-5):")
        for value, count in value_counts.head().items():
            print(f"    {value}: {count} ({count/len(data)*100:.1f}%)")
```

## 4. Выявление аномалий (выбросов)

```python
# Функция для определения выбросов по методу IQR
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    return outliers, lower_bound, upper_bound

print("АНАЛИЗ ВЫБРОСОВ")
print("="*50)

outlier_analysis = {}
for col in numeric_cols:
    if col not in ['person_id', 'team_id']:
        outliers, lower, upper = detect_outliers_iqr(data, col)
        outlier_count = len(outliers)
        outlier_percent = (outlier_count / len(data)) * 100

        outlier_analysis[col] = {
            'count': outlier_count,
            'percent': outlier_percent,
            'lower_bound': lower,
            'upper_bound': upper
        }

        print(f"\n{col}:")
        print(f"  Выбросов: {outlier_count} ({outlier_percent:.1f}%)")
        print(f"  Границы: [{lower:.2f}, {upper:.2f}]")
        if outlier_count > 0:
            print(f"  Топ-5 выбросов: {sorted(outliers[col].tolist(), reverse=True)[:5]}")
```

```python
# 1. Визуализация выбросов с помощью box plots
numeric_cols_for_plot = [col for col in numeric_cols if col not in ['person_id', 'team_id']]

if len(numeric_cols_for_plot) > 0:
    fig, axes = plt.subplots(2, 2, figsize=(18, 12))
    axes = axes.ravel()

    for i, col in enumerate(numeric_cols_for_plot[:4]):
        if i < len(axes):
            axes[i].boxplot(data[col].dropna(), vert=True)
            axes[i].set_title(f'Box Plot для {col}')
            axes[i].set_ylabel('Значение')

            if col in outlier_analysis:
                outlier_info = outlier_analysis[col]
                axes[i].text(0.02, 0.98, f'Выбросов: {outlier_info["count"]} ({outlier_info["percent"]:.1f}%)',
                           transform=axes[i].transAxes, verticalalignment='top')


    plt.tight_layout()
    plt.show()
```

```python
# 4. Диаграмма рассеяния с пузырьками

plt.figure(figsize=(14, 8))

flag_num = data['greatest_75_flag'].map({'Y': 1, 'N': 0})

scatter = plt.scatter(
    data['from_year'],
    data['to_year'],
    s=data['season_exp'] * 20 + 30,
    c=flag_num,
    alpha=0.6,
    cmap='RdYlGn'
)

plt.title(
    'Жизненный цикл игроков NBA\n'
    '(X — начало карьеры, Y — конец карьеры,\n'
    'размер пузыря — продолжительность карьеры)'
)
plt.xlabel('Год начала карьеры (from_year)')
plt.ylabel('Год окончания карьеры (to_year)')

exp_sizes = [1, 5, 10, 15, 20]
labels = [
    '1 сезон',
    '5 сезонов',
    '10 сезонов',
    '15 сезонов',
    '20+ сезонов'
]

for exp, label in zip(exp_sizes, labels):
    plt.scatter([], [], s=exp * 20 + 30, c='gray', alpha=0.6, label=label)

plt.legend(title='Продолжительность карьеры', loc='upper left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
country_counts = data['country'].value_counts(dropna=True)

top_n = 5
top_countries = country_counts.head(top_n)
other_count = country_counts.iloc[top_n:].sum()

country_plot = pd.concat([top_countries, pd.Series({'Other': other_count})])

plt.figure(figsize=(10, 7))
plt.pie(
    country_plot.values,
    labels=country_plot.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Распределение игроков по странам (топ-5 + Other)')
plt.tight_layout()
plt.show()
```

```python
non_usa_data = data[data['country'] != 'USA']

country_counts = non_usa_data['country'].value_counts(dropna=True)

top_n = 10
top_countries = country_counts.head(top_n)
other_count = country_counts.iloc[top_n:].sum()

country_plot = pd.concat([top_countries, pd.Series({'Other': other_count})])

plt.figure(figsize=(10, 7))
plt.pie(
    country_plot.values,
    labels=country_plot.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Распределение игроков по странам (топ-10 без США + Other)')
plt.tight_layout()
plt.show()
```

```python
df = data.copy()

df = df.dropna(subset=['from_year'])
df['decade'] = (df['from_year'].astype(int) // 10) * 10

df['greatest_75_num'] = df['greatest_75_flag'].map({'Y': 1, 'N': 0}).fillna(0)

decade_stats = (
    df.groupby('decade')['greatest_75_num']
      .mean()
      .sort_index()
)

plt.figure(figsize=(12, 6))
plt.plot(decade_stats.index, decade_stats.values * 100, marker='o')
plt.title('Доля игроков Greatest 75 по эпохам')
plt.xlabel('Десятилетие начала карьеры')
plt.ylabel('Доля игроков (%)')
plt.grid(True, alpha=0.3)

for x, y in zip(decade_stats.index, decade_stats.values * 100):
    plt.text(x, y + 0.3, f'{y:.1f}%', ha='center')

plt.tight_layout()
plt.show()
```
